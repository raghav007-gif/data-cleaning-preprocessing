# Data Cleaning and Preprocessing

## Internship Task 01

This notebook focuses on cleaning and preprocessing a customer dataset using Python and Pandas.

## Objective

The objective is to identify and fix common data-quality problems such as missing values, duplicate records, inconsistent categorical values, income formatting, and date formats.

```python
import pandas as pd

# Load the raw dataset
df = pd.read_csv('../../data/raw_customer_data.csv')

df.head()
```

## Data Inspection

```python
print('Dataset shape:', df.shape)
print('\nMissing values:')
print(df.isnull().sum())
print('\nDuplicate Customer IDs:', df['Customer_ID'].duplicated().sum())
```

## Data Cleaning

```python
# Remove duplicate customer records
df = df.drop_duplicates(subset='Customer_ID')

# Standardize Gender values
df['Gender'] = df['Gender'].astype(str).str.strip().str.title()

# Standardize City values
df['City'] = df['City'].astype(str).str.strip().str.title()

# Clean Income values and convert to numeric
df['Income'] = df['Income'].astype(str).str.replace(',', '', regex=False)
df['Income'] = pd.to_numeric(df['Income'], errors='coerce')

# Convert Purchase Amount to numeric
df['Purchase_Amount'] = pd.to_numeric(df['Purchase_Amount'], errors='coerce')

# Convert Purchase Date to datetime
df['Purchase_Date'] = pd.to_datetime(df['Purchase_Date'], errors='coerce')

# Fill missing numerical values with median
for column in ['Age', 'Income', 'Purchase_Amount']:
    df[column] = df[column].fillna(df[column].median())

# Fill missing categorical values with mode
for column in ['Gender', 'City', 'Category']:
    df[column] = df[column].fillna(df[column].mode()[0])

df.head()
```

## Verification

```python
print('Missing values after cleaning:')
print(df.isnull().sum())

print('\nDuplicate Customer IDs after cleaning:')
print(df['Customer_ID'].duplicated().sum())
```

```python
# Save the cleaned dataset
df.to_csv('../../data/cleaned_customer_data.csv', index=False)
    
print('Cleaned dataset saved successfully.')
```
